```python
import gymnasium as gym
```

```python
env_ids = list(gym.envs.registry.keys())
print(f"Total number of environments: {len(env_ids)}")
```

```python
print("\nAvailable Environments")
for name in sorted(env_ids):
    print(name)
```

```python
environments = ["CartPole-v1", "FrozenLake-v1", "MountainCar-v0", "Blackjack-v1", "Taxi-v3", "CliffWalking-v1"]
```

```python
for env_id in environments:
    print(f"\n{env_id} ")
```

```python
 env = gym.make(env_id)
```

```python
print("\n--- MDP Components Defined ---")
if env_id == "CartPole-v1":
        print("State:       Continuous vector [Cart Position, Cart Velocity, Pole Angle, Pole Angular Velocity]")
        print("Action:      Discrete [0: Push Left, 1: Push Right]")
        print("Transition:  Deterministic physics equations of motion")
        print("Reward:      +1.0 for every timestep the pole stays upright")

elif env_id == "FrozenLake-v1":
        print("State:       Discrete index (0 to 15) for a 4x4 grid")
        print("Action:      Discrete [0: Left, 1: Down, 2: Right, 3: Up]")
        print("Transition:  Stochastic (1/3 chance of sliding to unintended sides)")
        print("Reward:      +1.0 for reaching the Goal tile, 0 otherwise")

elif env_id == "MountainCar-v0":
        print("State:       Continuous vector [Car Position, Car Velocity]")
        print("Action:      Discrete [0: Accelerate Left, 1: Neutral, 2: Accelerate Right]")
        print("Transition:  Deterministic physics involving gravity and engine force")
        print("Reward:      -1.0 for every timestep until the target flag is reached")

elif env_id == "Blackjack-v1":
        print("State:       Tuple [Player Hand Sum, Dealer Showing Card, Usable Ace Boolean]")
        print("Action:      Discrete [0: Stick, 1: Hit]")
        print("Transition:  Stochastic depending on randomly drawn cards")
        print("Reward:      +1.0 for a win, 0.0 for a draw, -1.0 for a loss")

elif env_id == "Taxi-v3":
        print("State:       Discrete index representing taxi grid location, passenger index, and destination")
        print("Action:      Discrete [0-3: Move directions, 4: Pick up, 5: Drop off]")
        print("Transition:  Deterministic grid movements blocked by walls")
        print("Reward:      -1 per step, +20 for successful drop-off, -10 for illegal action")

elif env_id == "CliffWalking-v1":
        print("State:       Discrete index (0 to 47) for a 4x12 grid")
        print("Action:      Discrete [0: Up, 1: Right, 2: Down, 3: Left]")
        print("Transition:  Deterministic grid movement; stepping into the cliff resets you to start")
        print("Reward:      -1 per step, -100 for stepping into the cliff")

env.close()
```
